## VALIDAÇÃO DAS REGRAS CANDIDATAS PARA V3

Esta seção avalia cada regra sem alterar a V2 e sem criar a V3. As regras R1–R5 entram apenas em uma união hipotética; R6 permanece exclusivamente diagnóstica.

- **R1:** `Preco_m2 < 500`.
- **R2:** `Base de Cálculo adotada == 0`.
- **R3:** valor de transação dividido pela base de cálculo adotada `< 0.2`; bases iguais a zero ou ausentes não recebem razão e são tratadas apenas por R2.
- **R4:** data de transação fora de 2025; datas inválidas ou ausentes também ficariam fora do recorte.
- **R5:** cópias excedentes de linhas exatamente iguais, preservando a primeira ocorrência. O total de participantes dos grupos duplicados também é informado.
- **R6:** possíveis duplicatas por chaves de negócio. Não integra a união e não implica remoção.

### Significado da Base de Cálculo adotada

No ITBI paulistano, a base de cálculo tem natureza **fiscal**: o art. 7º da [Lei Municipal nº 11.154/1991](https://legislacao.prefeitura.sp.gov.br/lei-11154-de-30-de-dezembro-de-1991) a define como o valor venal do bem ou direito transmitido, entendido como o valor de negociação à vista em condições normais de mercado. A própria legislação também prevê situações e procedimentos específicos de arbitramento ou avaliação.

Portanto, `Base de Cálculo adotada` não deve ser tratada automaticamente como verdade observada do preço de mercado nem como uma medição independente do valor declarado. R2 e R3 são sinais de auditoria da coerência administrativa/econômica do registro. A célula seguinte verifica o tipo, a completude e a distribuição empírica antes de construir essas regras.

```python
# ============================================================
# AUDITORIA DA COLUNA BASE DE CÁLCULO ADOTADA
# ============================================================

col_base_calculo = "Base de Cálculo adotada"
base_calculo_original = df_v2[col_base_calculo]
base_calculo_v2 = pd.to_numeric(base_calculo_original, errors="coerce")
vvr_v2 = pd.to_numeric(df_v2["Valor Venal de Referência"], errors="coerce")

auditoria_base_calculo = pd.DataFrame(
    {
        "Métrica": [
            "dtype original",
            "registros",
            "ausentes originais",
            "falhas adicionais na conversão numérica",
            "valores iguais a zero",
            "valores negativos",
            "valores positivos",
            "valores únicos",
        ],
        "Valor": [
            str(base_calculo_original.dtype),
            len(base_calculo_original),
            int(base_calculo_original.isna().sum()),
            int((base_calculo_original.notna() & base_calculo_v2.isna()).sum()),
            int(base_calculo_v2.eq(0).sum()),
            int(base_calculo_v2.lt(0).sum()),
            int(base_calculo_v2.gt(0).sum()),
            int(base_calculo_v2.nunique(dropna=True)),
        ],
    }
)
display(auditoria_base_calculo)

print("Distribuição numérica da base de cálculo adotada:")
display(
    base_calculo_v2.describe(
        percentiles=[0.01, 0.05, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
    ).to_frame("Base de Cálculo adotada")
)

faixas_base_calculo = pd.cut(
    base_calculo_v2,
    bins=[-np.inf, 0, 10_000, 50_000, 100_000, 250_000, 500_000, 1_000_000, np.inf],
    include_lowest=True,
)
distribuicao_faixas_base = (
    faixas_base_calculo.value_counts(sort=False, dropna=False)
    .rename("Registros")
    .to_frame()
)
distribuicao_faixas_base["Percentual_V2"] = (
    distribuicao_faixas_base["Registros"] / len(df_v2) * 100
)
print("Distribuição por faixas:")
display(distribuicao_faixas_base)

comparacao_base = pd.DataFrame(
    {
        "Comparação": [
            "base adotada = valor venal de referência",
            "base adotada = valor de transação",
        ],
        "Registros": [
            int(np.isclose(base_calculo_v2, vvr_v2, rtol=0, atol=0.01, equal_nan=False).sum()),
            int(np.isclose(base_calculo_v2, df_v2[col_preco], rtol=0, atol=0.01, equal_nan=False).sum()),
        ],
    }
)
comparacao_base["Percentual_V2"] = comparacao_base["Registros"] / len(df_v2) * 100
print("Relação empírica com outras colunas monetárias:")
display(comparacao_base)

print("Valores mais frequentes da base de cálculo adotada:")
display(base_calculo_v2.value_counts(dropna=False).head(15).to_frame("Registros"))
```

```python
# ============================================================
# CONSTRUÇÃO DAS MÁSCARAS DIAGNÓSTICAS R1–R5
# ============================================================

data_transacao_v2 = pd.to_datetime(df_v2["Data de Transação"], errors="coerce")
razao_transacao_base_v2 = df_v2[col_preco] / base_calculo_v2.replace(0, np.nan)

mask_r1 = df_v2["Preco_m2"].lt(500)
mask_r2 = base_calculo_v2.eq(0)
mask_r3 = razao_transacao_base_v2.lt(0.2)
mask_r4 = data_transacao_v2.dt.year.ne(2025)
mask_r5 = df_v2.duplicated(keep="first")
mask_r5_grupos = df_v2.duplicated(keep=False)

mascaras_regras = pd.DataFrame(
    {"R1": mask_r1, "R2": mask_r2, "R3": mask_r3, "R4": mask_r4, "R5": mask_r5},
    index=df_v2.index,
)

df_validacao_regras = df_v2_analise.copy()
df_validacao_regras["Base_calculo_numerica"] = base_calculo_v2
df_validacao_regras["Razao_transacao_base"] = razao_transacao_base_v2
df_validacao_regras["Data_analise"] = data_transacao_v2

contagens_iniciais = mascaras_regras.sum().rename("Registros").to_frame()
contagens_iniciais["Percentual_V2"] = contagens_iniciais["Registros"] / len(df_v2) * 100
display(contagens_iniciais)

print("R5 — participantes de grupos de linhas exatamente iguais:")
print(f"{mask_r5_grupos.sum():,} registros")
print("R5 — cópias excedentes candidatas à remoção, mantendo uma ocorrência:")
print(f"{mask_r5.sum():,} registros")
```

```python
# ============================================================
# DIAGNÓSTICO INDIVIDUAL DE R1–R5
# ============================================================

colunas_exemplo_regras = [
    "N° do Cadastro (SQL)",
    col_preco,
    col_area,
    "Preco_m2",
    col_base_calculo,
    "Razao_transacao_base",
    "Data de Transação",
    "Mes_Arquivo",
    "CEP_8",
    "Matrícula do Imóvel",
]

ordenacao_exemplos = {
    "R1": ["Preco_m2"],
    "R2": ["Preco_m2"],
    "R3": ["Razao_transacao_base"],
    "R4": ["Data_analise"],
    "R5": ["N° do Cadastro (SQL)", "Data_analise"],
}

def exibir_diagnostico_regra(nome_regra, mascara, n_exemplos=10):
    parte = df_validacao_regras.loc[mascara].copy()
    percentual = len(parte) / len(df_v2) * 100

    print(f"\n{nome_regra}: {len(parte):,} registros ({percentual:.4f}% da V2)")
    estatisticas = parte[[col_preco, "Preco_m2", col_area]].describe(
        percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
    ).T
    display(estatisticas)

    distribuicao_ano = (
        parte["Data_analise"].dt.year
        .value_counts(dropna=False)
        .sort_index()
        .rename("Registros")
        .to_frame()
    )
    distribuicao_ano["Percentual_regra"] = distribuicao_ano["Registros"] / len(parte) * 100
    print("Distribuição por ano:")
    display(distribuicao_ano)

    distribuicao_mes = (
        parte["Data_analise"].dt.to_period("M")
        .value_counts(dropna=False)
        .sort_index()
        .rename("Registros")
        .to_frame()
    )
    distribuicao_mes["Percentual_regra"] = distribuicao_mes["Registros"] / len(parte) * 100
    print("Distribuição por mês de transação:")
    display(distribuicao_mes)

    print("Exemplos:")
    display(
        parte.sort_values(ordenacao_exemplos[nome_regra], na_position="last")
        [colunas_exemplo_regras]
        .head(n_exemplos)
    )

for regra in ["R1", "R2", "R3", "R4", "R5"]:
    exibir_diagnostico_regra(regra, mascaras_regras[regra])

print("R5 — exemplos com a ocorrência preservada e suas cópias exatas:")
display(
    df_validacao_regras.loc[mask_r5_grupos, colunas_exemplo_regras]
    .sort_values(["N° do Cadastro (SQL)", "Data de Transação", col_preco])
    .head(30)
)
```

### Interpretação individual das regras

- **R1 — preço/m² extremo:** é um sinal econômico forte e útil para revisão. Porém, deriva diretamente do alvo (`Valor de Transação`) e pode representar venda legítima com desconto, leilão, particularidade física não observada ou erro na área. Um corte automático introduziria seleção pelo próprio alvo; por isso, R1 é perigosa quando usada isoladamente.
- **R2 — base fiscal zero:** indica que a coluna fiscal não oferece referência monetária útil naquele registro, mas não prova que a transação declarada seja inválida. Pode refletir procedimento administrativo, decisão judicial, guia complementar ou ausência de valor de referência. R2 isolada é perigosa para exclusão.
- **R3 — razão inferior a 0,2:** evidencia grande divergência entre o valor declarado e a base fiscal adotada. Ainda assim, a base fiscal não é uma segunda observação independente do preço e pode incorporar regras administrativas. R3 deve servir como alerta de coerência, não como fundamento único de remoção.
- **R4 — data fora de 2025:** é uma incompatibilidade com o escopo temporal da base, não necessariamente uma transação anômala. Se o objetivo da V3 for representar transações ocorridas em 2025, R4 é defensável como regra de coorte; os registros continuam legítimos para outros recortes e devem permanecer preservados na fonte.
- **R5 — duplicata exata:** cópias integralmente iguais acrescentam peso artificial sem informação nova. Manter uma ocorrência por grupo é a regra candidata mais segura, desde que a ordem de ingestão usada para escolher a ocorrência preservada seja registrada.

```python
# ============================================================
# R6 — POSSÍVEIS DUPLICATAS POR CHAVES DE NEGÓCIO
# ============================================================

# Visão diagnóstica com apenas uma ocorrência de cada linha exata.
# Isto evita que R5 infle a investigação de R6; df_v2 não é alterado.
base_r6 = df_validacao_regras.loc[~mask_r5].copy()

estrategias_r6 = {
    "K1_CEP_area_valor_data": ["CEP_8", col_area, col_preco, "Data_analise"],
    "K2_SQL_area_valor_data": ["N° do Cadastro (SQL)", col_area, col_preco, "Data_analise"],
    "K3_cartorio_matricula_valor_data": ["Cartório de Registro", "Matrícula do Imóvel", col_preco, "Data_analise"],
    "K4_SQL_cartorio_matricula_valor_data": ["N° do Cadastro (SQL)", "Cartório de Registro", "Matrícula do Imóvel", col_preco, "Data_analise"],
}

resumo_r6 = []
flags_r6 = pd.DataFrame(index=base_r6.index)

for nome, chaves in estrategias_r6.items():
    flag = base_r6.duplicated(subset=chaves, keep=False)
    flags_r6[nome] = flag
    tamanhos = base_r6.groupby(chaves, dropna=False).size()
    grupos = tamanhos[tamanhos.gt(1)]
    resumo_r6.append(
        {
            "Estratégia": nome,
            "Registros_em_grupos": int(flag.sum()),
            "Grupos_candidatos": int(len(grupos)),
            "Excedentes_se_deduplicasse": int((grupos - 1).sum()),
            "Percentual_V2": flag.sum() / len(df_v2) * 100,
        }
    )

resumo_r6 = pd.DataFrame(resumo_r6)
display(resumo_r6)

print("Combinações de estratégias R6 satisfeitas pelos registros candidatos:")
display(
    flags_r6.loc[flags_r6.any(axis=1)]
    .value_counts()
    .rename("Registros")
    .reset_index()
)

colunas_exemplo_r6 = [
    "N° do Cadastro (SQL)",
    "Cartório de Registro",
    "Matrícula do Imóvel",
    "CEP_8",
    col_area,
    col_preco,
    "Data de Transação",
    "Mes_Arquivo",
    "Bairro",
]

for nome, chaves in estrategias_r6.items():
    print(f"Exemplos — {nome}:")
    display(
        base_r6.loc[flags_r6[nome]]
        .sort_values(chaves)[colunas_exemplo_r6]
        .head(16)
    )

print("R6 não foi incorporada a nenhuma máscara de remoção.")
```

### Leitura de R6

As chaves têm forças diferentes. `CEP + área + valor + data` é deliberadamente ampla e pode reunir unidades distintas do mesmo empreendimento vendidas nas mesmas condições. `SQL` é mais discriminante, enquanto `cartório + matrícula` se aproxima mais da identidade registral do imóvel. Mesmo as chaves fortes não identificam, por si sós, qual linha deveria ser descartada e podem representar guias complementares ou eventos administrativos distintos.

Por isso, os números de “excedentes se deduplicasse” são apenas uma medida de impacto potencial. R6 não deve remover registros até existir uma chave transacional estável ou uma verificação adicional de campos administrativos que diferenciem as ocorrências.

```python
# ============================================================
# SOBREPOSIÇÃO ENTRE R1–R5
# ============================================================

matriz_sobreposicao = mascaras_regras.astype("int64").T.dot(
    mascaras_regras.astype("int64")
)
print("Matriz de interseção — diagonal = total da regra; fora da diagonal = interseção do par:")
display(matriz_sobreposicao)

print(f"R1 também em R3: {(mask_r1 & mask_r3).sum():,}")
print(f"R1 também em R2: {(mask_r1 & mask_r2).sum():,}")

uniao_regras = mascaras_regras.any(axis=1)
quantidade_regras_por_registro = mascaras_regras.sum(axis=1)

interpretacoes_regras = {
    "R1": "Sinal econômico forte, mas deriva do alvo; perigosa isoladamente.",
    "R2": "Base fiscal ausente em substância; não invalida a venda por si só.",
    "R3": "Divergência fiscal elevada; usar como alerta, não como prova.",
    "R4": "Regra de escopo temporal, não de qualidade intrínseca.",
    "R5": "Cópia integral redundante; manter uma ocorrência.",
}

linhas_resumo = []
for regra in mascaras_regras.columns:
    mascara = mascaras_regras[regra]
    exclusivos = mascara & quantidade_regras_por_registro.eq(1)
    sobreposicao = mascara & quantidade_regras_por_registro.gt(1)
    linhas_resumo.append(
        {
            "Regra": regra,
            "Registros": int(mascara.sum()),
            "% V2": mascara.mean() * 100,
            "Exclusivos": int(exclusivos.sum()),
            "Sobreposição": int(sobreposicao.sum()),
            "Interpretação": interpretacoes_regras[regra],
        }
    )

resumo_regras_v3 = pd.DataFrame(linhas_resumo)
display(resumo_regras_v3)

print("Combinações observadas entre as regras, excluindo quem não satisfaz nenhuma:")
display(
    mascaras_regras.loc[uniao_regras]
    .value_counts()
    .rename("Registros")
    .reset_index()
)

print("Os exclusivos são a contribuição marginal de cada regra à união completa, sem depender de ordem.")
```

```python
# ============================================================
# IMPACTO HIPOTÉTICO DA UNIÃO — SEM CRIAR OU FILTRAR A V3
# ============================================================

n_v2_atual = len(df_v2)
n_candidatos_uniao = int(uniao_regras.sum())
n_pos_uniao_hipotetico = n_v2_atual - n_candidatos_uniao

impacto_hipotetico = pd.DataFrame(
    {
        "Métrica": [
            "V2 atual",
            "Registros candidatos à remoção pela união R1 | R2 | R3 | R4 | R5",
            "V3 hipotética (somente contagem)",
            "Percentual total preservado desde V1",
            "Percentual total preservado desde a base original de apartamentos",
        ],
        "Valor": [
            n_v2_atual,
            n_candidatos_uniao,
            n_pos_uniao_hipotetico,
            n_pos_uniao_hipotetico / len(df_v1) * 100,
            n_pos_uniao_hipotetico / len(df_apartamentos) * 100,
        ],
    }
)
display(impacto_hipotetico)

print("Nenhum registro foi removido e nenhum objeto df_v3 foi criado.")
```

### Recomendação metodológica antes da V3

A soma simples de R1–R5 seria 1.281, mas a união correta contém **1.098 registros**. Se todas as regras fossem aplicadas, restariam hipoteticamente 62.783 registros: 96,66% da V1 e 79,36% da base original de apartamentos. Essa união integral **não é recomendada**.

A evidência atual sustenta a seguinte decisão:

1. **R5 deve compor a V3**, removendo apenas as 48 cópias excedentes e preservando uma ocorrência de cada linha exata. É a regra com menor risco de eliminar informação real.
2. **R4 deve compor a V3 somente se a população-alvo for definida por `Data de Transação` em 2025.** Nesse caso, seus 669 registros são excluídos por escopo, não por serem inválidos. Como R4 e R5 não se sobrepõem, a aplicação somente dessas duas regras deixaria 63.164 registros, preservando 97,25% da V1 e 79,85% da base original de apartamentos. Se o objetivo for estudar guias pagas em 2025 independentemente da data da transação, R4 não deve ser aplicada.
3. **R1, R2 e R3 não devem compor a V3 como regras isoladas.** R1 deriva do próprio alvo e pode apagar vendas legítimas da cauda inferior. R2 não acrescenta nenhum registro exclusivo: seus 14 casos já pertencem a R1. R3 acrescenta 188 casos exclusivos e usa uma base fiscal que coincide com o próprio valor declarado em 79,86% da V2, portanto não funciona como validação independente do preço.
4. A interseção **`R1 & (R2 | R3)` reúne 168 registros** e é mais defensável como fila de revisão ou quarentena do que qualquer uma das três regras isoladas, pois combina preço/m² extremo com um segundo sinal fiscal. Ainda assim, não deve virar remoção automática antes de validar amostras na fonte e entender os casos de base zero.
5. **R6 não deve compor a V3 neste momento.** A chave ampla `CEP + área + valor + data` reuniu 1.262 registros e captura claramente vendas distintas de unidades semelhantes. As chaves mais fortes `SQL + área + valor + data` e `SQL + cartório + matrícula + valor + data` apontaram somente 22 registros em 11 grupos, mas ainda podem representar republicações ou eventos administrativos legítimos; falta uma chave única da guia/transação para decidir qual ocorrência seria redundante.

As regras mais perigosas para transações legítimas são **R1 e R3**, seguidas por R2. R4 também remove transações legítimas, mas de forma transparente por definição de coorte. A recomendação operacional, antes de criar a V3, é confirmar formalmente se o recorte temporal desejado é a data da transação; com essa confirmação, aplicar apenas `R4 | R5` e manter R1–R3 e R6 como flags auditáveis.

## BASE V3 - APLICAÇÃO FORMAL DE R4 E R5

A V3 aplica somente as regras aprovadas, nesta ordem:

1. **R4:** manter registros cuja `Data de Transação` pertença ao ano de 2025. Esta é uma regra de coorte temporal, não uma declaração de que as demais transações sejam inválidas.
2. **R5:** na base resultante de R4, remover apenas cópias excedentes de linhas exatamente iguais, mantendo deterministicamente a primeira ocorrência segundo a ordem original da V2.

R1, R2 e R3 não são filtros. A interseção `R1 & (R2 | R3)` será mantida como flag de auditoria sempre que o registro também satisfizer R4 e não for a cópia excedente removida por R5. `df_v1` e `df_v2` não serão alterados.

```python
# ============================================================
# VALIDAÇÃO SEQUENCIAL E CRIAÇÃO DA V3
# ============================================================

n_v1_antes_v3 = len(df_v1)
n_v2_antes_v3 = len(df_v2)
colunas_originais_v2 = df_v2.columns.tolist()

base_v3_trabalho = df_v2.copy()
base_v3_trabalho["_Ordem_original_V2"] = np.arange(len(base_v3_trabalho), dtype="int64")
data_v3_trabalho = pd.to_datetime(base_v3_trabalho["Data de Transação"], errors="coerce")

# R4: coorte definida pela data da transação.
mask_manter_r4 = data_v3_trabalho.dt.year.eq(2025)
n_removidos_r4 = int((~mask_manter_r4).sum())
base_pos_r4 = base_v3_trabalho.loc[mask_manter_r4].copy()

# R5: a ordenação estável torna explícita a ocorrência preservada.
base_pos_r4 = base_pos_r4.sort_values("_Ordem_original_V2", kind="mergesort")
mask_remover_r5_pos_r4 = base_pos_r4.duplicated(
    subset=colunas_originais_v2,
    keep="first",
)
n_removidos_r5_pos_r4 = int(mask_remover_r5_pos_r4.sum())
n_v3_confirmado = len(base_pos_r4) - n_removidos_r5_pos_r4

validacao_previa_v3 = pd.DataFrame(
    {
        "Etapa": ["V2 inicial", "Após R4", "R5 efetivo após R4", "V3 esperada"],
        "Registros": [len(df_v2), len(base_pos_r4), n_removidos_r5_pos_r4, n_v3_confirmado],
    }
)
display(validacao_previa_v3)

assert n_removidos_r4 == 669, "R4 não reproduziu os 669 registros validados."
assert n_removidos_r5_pos_r4 == 48, "R5 após R4 não reproduziu as 48 cópias excedentes."
assert n_v3_confirmado == 63_164, "A aplicação sequencial de R4 e R5 não resultou em 63.164 registros."

indices_removidos_r5 = base_pos_r4.index[mask_remover_r5_pos_r4]
df_v3 = (
    base_pos_r4.loc[~mask_remover_r5_pos_r4]
    .sort_values("_Ordem_original_V2", kind="mergesort")
    .drop(columns="_Ordem_original_V2")
    .copy()
)

assert len(df_v3) == 63_164
assert len(df_v1) == n_v1_antes_v3
assert len(df_v2) == n_v2_antes_v3

print("df_v3 criada com sucesso, após validação das contagens.")
print(f"Registros: {len(df_v3):,}")
```

```python
# ============================================================
# FLAG E BASE DE AUDITORIA PARA R1 & (R2 | R3)
# ============================================================

base_calculo_auditoria = pd.to_numeric(df_v2[col_base_calculo], errors="coerce")
razao_auditoria = df_v2[col_preco] / base_calculo_auditoria.replace(0, np.nan)
mask_auditoria_baixo_m2 = (
    df_v2["Preco_m2"].lt(500)
    & (base_calculo_auditoria.eq(0) | razao_auditoria.lt(0.2))
)

assert mask_auditoria_baixo_m2.sum() == 168
indices_auditoria = df_v2.index[mask_auditoria_baixo_m2]

df_auditoria_baixo_m2 = df_v2.loc[mask_auditoria_baixo_m2].copy()
df_auditoria_baixo_m2["Flag_R1"] = True
df_auditoria_baixo_m2["Flag_R2"] = base_calculo_auditoria.loc[mask_auditoria_baixo_m2].eq(0)
df_auditoria_baixo_m2["Flag_R3"] = razao_auditoria.loc[mask_auditoria_baixo_m2].lt(0.2)
df_auditoria_baixo_m2["Razao_transacao_base"] = razao_auditoria.loc[mask_auditoria_baixo_m2]
df_auditoria_baixo_m2["Presente_na_V3"] = df_auditoria_baixo_m2.index.isin(df_v3.index)
df_auditoria_baixo_m2["Excluido_por_R4"] = (
    pd.to_datetime(df_auditoria_baixo_m2["Data de Transação"], errors="coerce")
    .dt.year.ne(2025)
)
df_auditoria_baixo_m2["Excluido_por_R5"] = df_auditoria_baixo_m2.index.isin(indices_removidos_r5)
df_auditoria_baixo_m2["Situação_V3"] = np.select(
    [
        df_auditoria_baixo_m2["Excluido_por_R4"],
        df_auditoria_baixo_m2["Excluido_por_R5"],
    ],
    ["Fora da coorte por R4", "Cópia excedente removida por R5"],
    default="Presente na V3 e marcado para auditoria",
)

df_v3["Flag_auditoria_baixo_m2"] = df_v3.index.isin(indices_auditoria)

assert len(df_auditoria_baixo_m2) == 168
assert df_v3["Flag_auditoria_baixo_m2"].sum() == 161

print("Situação dos 168 registros da auditoria:")
display(df_auditoria_baixo_m2["Situação_V3"].value_counts().to_frame("Registros"))

print("Exemplos da base de auditoria:")
display(
    df_auditoria_baixo_m2[
        [
            col_preco, col_area, "Preco_m2", col_base_calculo,
            "Razao_transacao_base", "Data de Transação",
            "Presente_na_V3", "Situação_V3",
        ]
    ].sort_values("Preco_m2").head(20)
)
```

```python
# ============================================================
# RELATÓRIO DE VALIDAÇÃO DA V3
# ============================================================

datas_v3 = pd.to_datetime(df_v3["Data de Transação"], errors="coerce")
todas_datas_2025 = bool(datas_v3.notna().all() and datas_v3.dt.year.eq(2025).all())
duplicatas_exatas_restantes = int(
    df_v3.duplicated(subset=colunas_originais_v2, keep="first").sum()
)

relatorio_v3 = pd.DataFrame(
    {
        "Métrica": [
            "V1",
            "V2",
            "V3",
            "Removidos por R4",
            "Removidos efetivamente por R5 após R4",
            "Percentual preservado da V1",
            "Percentual preservado da base original de apartamentos",
            "Todas as datas da V3 pertencem a 2025",
            "Duplicatas exatas excedentes restantes",
            "Registros marcados para auditoria na V3",
        ],
        "Valor": [
            len(df_v1),
            len(df_v2),
            len(df_v3),
            n_removidos_r4,
            n_removidos_r5_pos_r4,
            len(df_v3) / len(df_v1) * 100,
            len(df_v3) / len(df_apartamentos) * 100,
            todas_datas_2025,
            duplicatas_exatas_restantes,
            int(df_v3["Flag_auditoria_baixo_m2"].sum()),
        ],
    }
)
display(relatorio_v3)

print("Intervalo de datas da V3:")
print(datas_v3.min(), "a", datas_v3.max())

print("Estatísticas das variáveis centrais na V3:")
estatisticas_v3 = df_v3[[col_preco, col_area, "Preco_m2"]].describe(
    percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
).T
display(estatisticas_v3)

assert todas_datas_2025
assert duplicatas_exatas_restantes == 0
assert len(df_v1) == n_v1_antes_v3 and len(df_v2) == n_v2_antes_v3
```

## INVESTIGAÇÃO DA CAUDA SUPERIOR DA V3

Esta seção é exclusivamente descritiva. Percentis identificam regiões da distribuição para inspeção, mas não constituem regras automáticas de exclusão. Nenhum IQR, z-score ou corte por percentil será usado para remover registros, e nenhuma V4 será criada.

```python
# ============================================================
# PERCENTIS E GRUPOS DESCRITIVOS DA CAUDA SUPERIOR
# ============================================================

metricas_cauda_superior = [col_preco, "Preco_m2", col_area]
percentis_cauda = [0.95, 0.97, 0.99, 0.995, 0.999]
rotulos_percentis = ["P95", "P97", "P99", "P99.5", "P99.9"]

limites_cauda_superior = (
    df_v3[metricas_cauda_superior]
    .quantile(percentis_cauda)
    .T
)
limites_cauda_superior.columns = rotulos_percentis
display(limites_cauda_superior)

contagens_acima_percentis = []
for metrica in metricas_cauda_superior:
    for rotulo in rotulos_percentis:
        limite = limites_cauda_superior.loc[metrica, rotulo]
        quantidade = int(df_v3[metrica].ge(limite).sum())
        contagens_acima_percentis.append(
            {
                "Métrica": metrica,
                "Percentil": rotulo,
                "Limite": limite,
                "Registros_no_ou_acima_do_limite": quantidade,
                "Percentual_V3": quantidade / len(df_v3) * 100,
            }
        )
display(pd.DataFrame(contagens_acima_percentis))

df_cauda_superior = df_v3.copy()
cep_num_cauda = pd.to_numeric(df_cauda_superior["CEP"], errors="coerce")
df_cauda_superior["CEP_8"] = cep_num_cauda.round().astype("Int64").astype("string").str.zfill(8)
df_cauda_superior.loc[cep_num_cauda.isna(), "CEP_8"] = pd.NA

df_cauda_superior["Extremo_valor_P995"] = df_cauda_superior[col_preco].ge(
    limites_cauda_superior.loc[col_preco, "P99.5"]
)
df_cauda_superior["Extremo_Preco_m2_P995"] = df_cauda_superior["Preco_m2"].ge(
    limites_cauda_superior.loc["Preco_m2", "P99.5"]
)
df_cauda_superior["Extremo_area_P995"] = df_cauda_superior[col_area].ge(
    limites_cauda_superior.loc[col_area, "P99.5"]
)

flags_extremos_p995 = ["Extremo_valor_P995", "Extremo_Preco_m2_P995", "Extremo_area_P995"]
df_cauda_superior["Qualquer_extremo_P995"] = df_cauda_superior[flags_extremos_p995].any(axis=1)
df_cauda_superior["Grupo_extremo_P995"] = df_cauda_superior[flags_extremos_p995].apply(
    lambda linha: " + ".join(
        nome for nome, ativo in zip(["Valor", "Preco_m2", "Area"], linha) if ativo
    ) or "Fora da cauda P99.5",
    axis=1,
)

print("Grupos formados pela interseção dos três percentis 99,5:")
display(df_cauda_superior["Grupo_extremo_P995"].value_counts().to_frame("Registros"))
```

```python
# ============================================================
# RELAÇÃO ENTRE PREÇO, ÁREA E PREÇO/M²
# ============================================================

print("Correlação de Pearson:")
display(df_v3[metricas_cauda_superior].corr(method="pearson"))
print("Correlação de Spearman:")
display(df_v3[metricas_cauda_superior].corr(method="spearman"))

resumo_grupos_extremos = (
    df_cauda_superior[df_cauda_superior["Qualquer_extremo_P995"]]
    .groupby("Grupo_extremo_P995")
    .agg(
        Registros=(col_preco, "size"),
        Valor_mediano=(col_preco, "median"),
        Valor_maximo=(col_preco, "max"),
        Area_mediana=(col_area, "median"),
        Area_maxima=(col_area, "max"),
        Preco_m2_mediano=("Preco_m2", "median"),
        Preco_m2_maximo=("Preco_m2", "max"),
    )
    .sort_values("Registros", ascending=False)
)
display(resumo_grupos_extremos)

p5_area = df_v3[col_area].quantile(0.05)
p5_preco_m2 = df_v3["Preco_m2"].quantile(0.05)
mediana_valor = df_v3[col_preco].median()

sinais_revisao_superior = {
    "Luxo_consistente_valor_e_preco_m2": (
        df_cauda_superior["Extremo_valor_P995"]
        & df_cauda_superior["Extremo_Preco_m2_P995"]
    ),
    "Imovel_grande_e_valioso": (
        df_cauda_superior["Extremo_valor_P995"]
        & df_cauda_superior["Extremo_area_P995"]
    ),
    "Revisar_area_possivelmente_baixa": (
        df_cauda_superior["Extremo_Preco_m2_P995"]
        & df_cauda_superior[col_area].le(p5_area)
        & df_cauda_superior[col_preco].lt(mediana_valor)
    ),
    "Revisar_area_possivelmente_alta": (
        df_cauda_superior["Extremo_area_P995"]
        & df_cauda_superior["Preco_m2"].le(p5_preco_m2)
    ),
    "Valor_extremo_sem_area_ou_preco_m2_extremos": (
        df_cauda_superior["Extremo_valor_P995"]
        & ~df_cauda_superior["Extremo_Preco_m2_P995"]
        & ~df_cauda_superior["Extremo_area_P995"]
    ),
}

resumo_sinais_superiores = pd.DataFrame(
    [
        {
            "Perfil_ou_sinal": nome,
            "Registros": int(mascara.sum()),
            "Percentual_V3": mascara.mean() * 100,
        }
        for nome, mascara in sinais_revisao_superior.items()
    ]
)
display(resumo_sinais_superiores)

for nome, mascara in sinais_revisao_superior.items():
    print(f"Exemplos - {nome}:")
    display(
        df_cauda_superior.loc[mascara]
        .sort_values(col_preco, ascending=False)[colunas_exibicao_cauda]
        .head(12)
    )
```

```python
# ============================================================
# CONTEXTO GEOGRÁFICO, IPTU, TEMPORAL E FREQUÊNCIA
# ============================================================

grupos_metricas_superiores = {
    "Valor_P995": (df_cauda_superior["Extremo_valor_P995"], col_preco),
    "Preco_m2_P995": (df_cauda_superior["Extremo_Preco_m2_P995"], "Preco_m2"),
    "Area_P995": (df_cauda_superior["Extremo_area_P995"], col_area),
}

linhas_contexto_geo = []
linhas_contexto_padrao = []
linhas_frequencia_valores = []
distribuicao_mensal_extremos = {}

for grupo, (mascara, metrica) in grupos_metricas_superiores.items():
    parte = df_cauda_superior.loc[mascara]

    for campo in ["CEP_8", "Bairro"]:
        contagens = parte[campo].fillna("<ausente>").astype(str).value_counts().head(15)
        for valor, quantidade in contagens.items():
            linhas_contexto_geo.append(
                {"Grupo": grupo, "Campo": campo, "Valor": valor, "Registros": quantidade}
            )

    padrao = (
        parte["Padrão (IPTU)"].astype("string").fillna("<ausente>")
        + " | "
        + parte["Descrição do padrão (IPTU)"].astype("string").fillna("<ausente>")
    )
    for valor, quantidade in padrao.value_counts().head(15).items():
        linhas_contexto_padrao.append(
            {"Grupo": grupo, "Padrão_IPTU": valor, "Registros": quantidade}
        )

    distribuicao_mensal_extremos[grupo] = (
        pd.to_datetime(parte["Data de Transação"], errors="coerce")
        .dt.to_period("M")
        .value_counts()
        .sort_index()
    )

    for valor, quantidade in parte[metrica].value_counts().head(15).items():
        linhas_frequencia_valores.append(
            {"Grupo": grupo, "Métrica": metrica, "Valor": valor, "Frequência": quantidade}
        )

print("CEPs e bairros mais frequentes em cada grupo extremo:")
display(pd.DataFrame(linhas_contexto_geo))
print("Padrões IPTU mais frequentes em cada grupo extremo:")
display(pd.DataFrame(linhas_contexto_padrao))
print("Distribuição mensal dos grupos extremos:")
display(pd.DataFrame(distribuicao_mensal_extremos).fillna(0).astype(int))
print("Valores exatos mais repetidos dentro de cada grupo extremo:")
display(pd.DataFrame(linhas_frequencia_valores))
```

```python
# ============================================================
# COERÊNCIA FISCAL E ADMINISTRATIVA DOS GRUPOS EXTREMOS
# ============================================================

resumo_coerencia_superior = []
situacao_sql_superior = []

for grupo, (mascara, metrica) in grupos_metricas_superiores.items():
    parte = df_cauda_superior.loc[mascara]
    base_num = pd.to_numeric(parte[col_base_calculo], errors="coerce")
    vvr_num = pd.to_numeric(parte["Valor Venal de Referência"], errors="coerce")
    razao_valor_base = parte[col_preco] / base_num.replace(0, np.nan)

    resumo_coerencia_superior.append(
        {
            "Grupo": grupo,
            "Registros": len(parte),
            "Base_zero": int(base_num.eq(0).sum()),
            "Base_igual_valor": int(
                np.isclose(base_num, parte[col_preco], rtol=0, atol=0.01, equal_nan=False).sum()
            ),
            "Percentual_base_igual_valor": (
                np.isclose(base_num, parte[col_preco], rtol=0, atol=0.01, equal_nan=False).mean() * 100
            ),
            "Base_igual_VVR": int(
                np.isclose(base_num, vvr_num, rtol=0, atol=0.01, equal_nan=False).sum()
            ),
            "Razao_valor_base_mediana": razao_valor_base.median(),
            "Area_minima": parte[col_area].min(),
            "Area_maxima": parte[col_area].max(),
        }
    )

    for situacao, quantidade in parte["Situação do SQL"].fillna("<ausente>").value_counts().items():
        situacao_sql_superior.append(
            {"Grupo": grupo, "Situação_SQL": situacao, "Registros": quantidade}
        )

display(pd.DataFrame(resumo_coerencia_superior))
print("Situação cadastral do SQL nos grupos extremos:")
display(pd.DataFrame(situacao_sql_superior))

print("Checagens estruturais herdadas da definição da V1:")
print("Naturezas de transação na V3:", df_v3["Natureza de Transação"].unique())
print("Proporções transmitidas na V3:", df_v3["Proporção Transmitida (%)"].unique())
print("Áreas não positivas na V3:", int(df_v3[col_area].le(0).sum()))
print("Valores de transação não positivos na V3:", int(df_v3[col_preco].le(0).sum()))
```

### Grupos encontrados e recomendação metodológica

A região P99,5 de pelo menos uma das três métricas contém **639 registros**. Os grupos não são classes definitivas de qualidade; são perfis para orientar inspeção:

- **Valor + Preco_m2 + Area: 69 registros.** Reúnem preço total, preço/m² e área muito altos.
- **Valor + Preco_m2: 106 registros.** Somados ao grupo anterior, formam 175 casos com perfil forte de alto luxo. Os maiores valores se concentram em endereços e CEPs reconhecidamente valorizados, com padrões IPTU 24/25 e coerência entre valor, área e preço/m².
- **Valor + Area: 67 registros.** Somados ao primeiro grupo, formam 136 imóveis simultaneamente grandes e valiosos, também compatíveis com alto padrão legítimo.
- **Valor apenas: 74 registros.** Têm valor mediano de aproximadamente R$ 8,41 milhões, área mediana de 523 m² e preço/m² mediano de R$ 16,93 mil. Em geral, estão logo abaixo dos outros limiares e não apresentam sinal isolado de erro.
- **Preco_m2 apenas: 141 registros.** Têm área mediana de 55 m², valor mediano de aproximadamente R$ 1,73 milhão e preço/m² mediano de R$ 23,08 mil. O padrão é compatível com unidades compactas em localizações premium e vendas padronizadas de empreendimentos. A busca específica por preço/m² extremo combinado a área muito baixa e valor abaixo da mediana encontrou zero casos.
- **Area apenas: 182 registros.** Têm área mediana de 697,5 m² e preço/m² mediano de aproximadamente R$ 4,79 mil. Podem representar apartamentos antigos muito grandes, coberturas e unidades com vagas/depósitos incorporados, mas este é o grupo que mais merece verificação cadastral.

Há **35 registros** com área no P99,5 e preço/m² no P5 inferior. Eles não devem ser removidos: são candidatos à conferência da `Área Construída (m2)` no cadastro, especialmente quando a área supera 1.000 m² e o valor não acompanha proporcionalmente. A área máxima, 2.023 m², aparece em registro ativo, com valor de aproximadamente R$ 10,52 milhões e referências de vagas/depósito, portanto nem mesmo esse caso pode ser classificado como erro apenas pela magnitude.

Os dados administrativos reforçam a hipótese de legitimidade da maior parte da cauda: em cada um dos três recortes métricos P99,5, todos os registros têm situação `Ativo Predial`; não há base de cálculo zero; a base adotada coincide com o valor declarado em 99,37% do grupo de valor extremo e em 100% do grupo de preço/m² extremo. Todos continuam sendo compra e venda de 100% da proporção transmitida.

As possíveis duplicidades também não justificam remoção. A chave ampla `CEP + área + valor + data` alcança 62 registros da cauda P99,5, mas as chaves fortes com SQL, cartório e matrícula alcançam zero. Isso é compatível com unidades distintas de um mesmo empreendimento, não com cópias do mesmo registro.

**Recomendação:** preservar integralmente a cauda superior na V3. Não criar V4 e não aplicar corte automático por valor, área, preço/m², IQR, z-score ou percentil. Manter como prioridades de auditoria os 35 casos de área extrema com preço/m² baixo e, secundariamente, conferir amostras do P99,9 diretamente no cadastro. Para validações geográficas, preferir CEP, SQL e matrícula a `Bairro`, que permanece ruidoso. Os demais grupos apresentam evidência consistente com imóveis de luxo ou transações excepcionais legítimas.
