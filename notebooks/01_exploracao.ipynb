## VALIDAÇÃO DAS REGRAS CANDIDATAS PARA V3

Esta seção avalia cada regra sem alterar a V2 e sem criar a V3. As regras R1–R5 entram apenas em uma união hipotética; R6 permanece exclusivamente diagnóstica.

- **R1:** `Preco_m2 < 500`.
- **R2:** `Base de Cálculo adotada == 0`.
- **R3:** valor de transação dividido pela base de cálculo adotada `< 0.2`; bases iguais a zero ou ausentes não recebem razão e são tratadas apenas por R2.
- **R4:** data de transação fora de 2025; datas inválidas ou ausentes também ficariam fora do recorte.
- **R5:** cópias excedentes de linhas exatamente iguais, preservando a primeira ocorrência. O total de participantes dos grupos duplicados também é informado.
- **R6:** possíveis duplicatas por chaves de negócio. Não integra a união e não implica remoção.

### Significado da Base de Cálculo adotada

No ITBI paulistano, a base de cálculo tem natureza **fiscal**: o art. 7º da [Lei Municipal nº 11.154/1991](https://legislacao.prefeitura.sp.gov.br/lei-11154-de-30-de-dezembro-de-1991) a define como o valor venal do bem ou direito transmitido, entendido como o valor de negociação à vista em condições normais de mercado. A própria legislação também prevê situações e procedimentos específicos de arbitramento ou avaliação.

Portanto, `Base de Cálculo adotada` não deve ser tratada automaticamente como verdade observada do preço de mercado nem como uma medição independente do valor declarado. R2 e R3 são sinais de auditoria da coerência administrativa/econômica do registro. A célula seguinte verifica o tipo, a completude e a distribuição empírica antes de construir essas regras.

```python
# ============================================================
# AUDITORIA DA COLUNA BASE DE CÁLCULO ADOTADA
# ============================================================

col_base_calculo = "Base de Cálculo adotada"
base_calculo_original = df_v2[col_base_calculo]
base_calculo_v2 = pd.to_numeric(base_calculo_original, errors="coerce")
vvr_v2 = pd.to_numeric(df_v2["Valor Venal de Referência"], errors="coerce")

auditoria_base_calculo = pd.DataFrame(
    {
        "Métrica": [
            "dtype original",
            "registros",
            "ausentes originais",
            "falhas adicionais na conversão numérica",
            "valores iguais a zero",
            "valores negativos",
            "valores positivos",
            "valores únicos",
        ],
        "Valor": [
            str(base_calculo_original.dtype),
            len(base_calculo_original),
            int(base_calculo_original.isna().sum()),
            int((base_calculo_original.notna() & base_calculo_v2.isna()).sum()),
            int(base_calculo_v2.eq(0).sum()),
            int(base_calculo_v2.lt(0).sum()),
            int(base_calculo_v2.gt(0).sum()),
            int(base_calculo_v2.nunique(dropna=True)),
        ],
    }
)
display(auditoria_base_calculo)

print("Distribuição numérica da base de cálculo adotada:")
display(
    base_calculo_v2.describe(
        percentiles=[0.01, 0.05, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
    ).to_frame("Base de Cálculo adotada")
)

faixas_base_calculo = pd.cut(
    base_calculo_v2,
    bins=[-np.inf, 0, 10_000, 50_000, 100_000, 250_000, 500_000, 1_000_000, np.inf],
    include_lowest=True,
)
distribuicao_faixas_base = (
    faixas_base_calculo.value_counts(sort=False, dropna=False)
    .rename("Registros")
    .to_frame()
)
distribuicao_faixas_base["Percentual_V2"] = (
    distribuicao_faixas_base["Registros"] / len(df_v2) * 100
)
print("Distribuição por faixas:")
display(distribuicao_faixas_base)

comparacao_base = pd.DataFrame(
    {
        "Comparação": [
            "base adotada = valor venal de referência",
            "base adotada = valor de transação",
        ],
        "Registros": [
            int(np.isclose(base_calculo_v2, vvr_v2, rtol=0, atol=0.01, equal_nan=False).sum()),
            int(np.isclose(base_calculo_v2, df_v2[col_preco], rtol=0, atol=0.01, equal_nan=False).sum()),
        ],
    }
)
comparacao_base["Percentual_V2"] = comparacao_base["Registros"] / len(df_v2) * 100
print("Relação empírica com outras colunas monetárias:")
display(comparacao_base)

print("Valores mais frequentes da base de cálculo adotada:")
display(base_calculo_v2.value_counts(dropna=False).head(15).to_frame("Registros"))
```

```python
# ============================================================
# CONSTRUÇÃO DAS MÁSCARAS DIAGNÓSTICAS R1–R5
# ============================================================

data_transacao_v2 = pd.to_datetime(df_v2["Data de Transação"], errors="coerce")
razao_transacao_base_v2 = df_v2[col_preco] / base_calculo_v2.replace(0, np.nan)

mask_r1 = df_v2["Preco_m2"].lt(500)
mask_r2 = base_calculo_v2.eq(0)
mask_r3 = razao_transacao_base_v2.lt(0.2)
mask_r4 = data_transacao_v2.dt.year.ne(2025)
mask_r5 = df_v2.duplicated(keep="first")
mask_r5_grupos = df_v2.duplicated(keep=False)

mascaras_regras = pd.DataFrame(
    {"R1": mask_r1, "R2": mask_r2, "R3": mask_r3, "R4": mask_r4, "R5": mask_r5},
    index=df_v2.index,
)

df_validacao_regras = df_v2_analise.copy()
df_validacao_regras["Base_calculo_numerica"] = base_calculo_v2
df_validacao_regras["Razao_transacao_base"] = razao_transacao_base_v2
df_validacao_regras["Data_analise"] = data_transacao_v2

contagens_iniciais = mascaras_regras.sum().rename("Registros").to_frame()
contagens_iniciais["Percentual_V2"] = contagens_iniciais["Registros"] / len(df_v2) * 100
display(contagens_iniciais)

print("R5 — participantes de grupos de linhas exatamente iguais:")
print(f"{mask_r5_grupos.sum():,} registros")
print("R5 — cópias excedentes candidatas à remoção, mantendo uma ocorrência:")
print(f"{mask_r5.sum():,} registros")
```

```python
# ============================================================
# DIAGNÓSTICO INDIVIDUAL DE R1–R5
# ============================================================

colunas_exemplo_regras = [
    "N° do Cadastro (SQL)",
    col_preco,
    col_area,
    "Preco_m2",
    col_base_calculo,
    "Razao_transacao_base",
    "Data de Transação",
    "Mes_Arquivo",
    "CEP_8",
    "Matrícula do Imóvel",
]

ordenacao_exemplos = {
    "R1": ["Preco_m2"],
    "R2": ["Preco_m2"],
    "R3": ["Razao_transacao_base"],
    "R4": ["Data_analise"],
    "R5": ["N° do Cadastro (SQL)", "Data_analise"],
}

def exibir_diagnostico_regra(nome_regra, mascara, n_exemplos=10):
    parte = df_validacao_regras.loc[mascara].copy()
    percentual = len(parte) / len(df_v2) * 100

    print(f"\n{nome_regra}: {len(parte):,} registros ({percentual:.4f}% da V2)")
    estatisticas = parte[[col_preco, "Preco_m2", col_area]].describe(
        percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
    ).T
    display(estatisticas)

    distribuicao_ano = (
        parte["Data_analise"].dt.year
        .value_counts(dropna=False)
        .sort_index()
        .rename("Registros")
        .to_frame()
    )
    distribuicao_ano["Percentual_regra"] = distribuicao_ano["Registros"] / len(parte) * 100
    print("Distribuição por ano:")
    display(distribuicao_ano)

    distribuicao_mes = (
        parte["Data_analise"].dt.to_period("M")
        .value_counts(dropna=False)
        .sort_index()
        .rename("Registros")
        .to_frame()
    )
    distribuicao_mes["Percentual_regra"] = distribuicao_mes["Registros"] / len(parte) * 100
    print("Distribuição por mês de transação:")
    display(distribuicao_mes)

    print("Exemplos:")
    display(
        parte.sort_values(ordenacao_exemplos[nome_regra], na_position="last")
        [colunas_exemplo_regras]
        .head(n_exemplos)
    )

for regra in ["R1", "R2", "R3", "R4", "R5"]:
    exibir_diagnostico_regra(regra, mascaras_regras[regra])

print("R5 — exemplos com a ocorrência preservada e suas cópias exatas:")
display(
    df_validacao_regras.loc[mask_r5_grupos, colunas_exemplo_regras]
    .sort_values(["N° do Cadastro (SQL)", "Data de Transação", col_preco])
    .head(30)
)
```

### Interpretação individual das regras

- **R1 — preço/m² extremo:** é um sinal econômico forte e útil para revisão. Porém, deriva diretamente do alvo (`Valor de Transação`) e pode representar venda legítima com desconto, leilão, particularidade física não observada ou erro na área. Um corte automático introduziria seleção pelo próprio alvo; por isso, R1 é perigosa quando usada isoladamente.
- **R2 — base fiscal zero:** indica que a coluna fiscal não oferece referência monetária útil naquele registro, mas não prova que a transação declarada seja inválida. Pode refletir procedimento administrativo, decisão judicial, guia complementar ou ausência de valor de referência. R2 isolada é perigosa para exclusão.
- **R3 — razão inferior a 0,2:** evidencia grande divergência entre o valor declarado e a base fiscal adotada. Ainda assim, a base fiscal não é uma segunda observação independente do preço e pode incorporar regras administrativas. R3 deve servir como alerta de coerência, não como fundamento único de remoção.
- **R4 — data fora de 2025:** é uma incompatibilidade com o escopo temporal da base, não necessariamente uma transação anômala. Se o objetivo da V3 for representar transações ocorridas em 2025, R4 é defensável como regra de coorte; os registros continuam legítimos para outros recortes e devem permanecer preservados na fonte.
- **R5 — duplicata exata:** cópias integralmente iguais acrescentam peso artificial sem informação nova. Manter uma ocorrência por grupo é a regra candidata mais segura, desde que a ordem de ingestão usada para escolher a ocorrência preservada seja registrada.

```python
# ============================================================
# R6 — POSSÍVEIS DUPLICATAS POR CHAVES DE NEGÓCIO
# ============================================================

# Visão diagnóstica com apenas uma ocorrência de cada linha exata.
# Isto evita que R5 infle a investigação de R6; df_v2 não é alterado.
base_r6 = df_validacao_regras.loc[~mask_r5].copy()

estrategias_r6 = {
    "K1_CEP_area_valor_data": ["CEP_8", col_area, col_preco, "Data_analise"],
    "K2_SQL_area_valor_data": ["N° do Cadastro (SQL)", col_area, col_preco, "Data_analise"],
    "K3_cartorio_matricula_valor_data": ["Cartório de Registro", "Matrícula do Imóvel", col_preco, "Data_analise"],
    "K4_SQL_cartorio_matricula_valor_data": ["N° do Cadastro (SQL)", "Cartório de Registro", "Matrícula do Imóvel", col_preco, "Data_analise"],
}

resumo_r6 = []
flags_r6 = pd.DataFrame(index=base_r6.index)

for nome, chaves in estrategias_r6.items():
    flag = base_r6.duplicated(subset=chaves, keep=False)
    flags_r6[nome] = flag
    tamanhos = base_r6.groupby(chaves, dropna=False).size()
    grupos = tamanhos[tamanhos.gt(1)]
    resumo_r6.append(
        {
            "Estratégia": nome,
            "Registros_em_grupos": int(flag.sum()),
            "Grupos_candidatos": int(len(grupos)),
            "Excedentes_se_deduplicasse": int((grupos - 1).sum()),
            "Percentual_V2": flag.sum() / len(df_v2) * 100,
        }
    )

resumo_r6 = pd.DataFrame(resumo_r6)
display(resumo_r6)

print("Combinações de estratégias R6 satisfeitas pelos registros candidatos:")
display(
    flags_r6.loc[flags_r6.any(axis=1)]
    .value_counts()
    .rename("Registros")
    .reset_index()
)

colunas_exemplo_r6 = [
    "N° do Cadastro (SQL)",
    "Cartório de Registro",
    "Matrícula do Imóvel",
    "CEP_8",
    col_area,
    col_preco,
    "Data de Transação",
    "Mes_Arquivo",
    "Bairro",
]

for nome, chaves in estrategias_r6.items():
    print(f"Exemplos — {nome}:")
    display(
        base_r6.loc[flags_r6[nome]]
        .sort_values(chaves)[colunas_exemplo_r6]
        .head(16)
    )

print("R6 não foi incorporada a nenhuma máscara de remoção.")
```

### Leitura de R6

As chaves têm forças diferentes. `CEP + área + valor + data` é deliberadamente ampla e pode reunir unidades distintas do mesmo empreendimento vendidas nas mesmas condições. `SQL` é mais discriminante, enquanto `cartório + matrícula` se aproxima mais da identidade registral do imóvel. Mesmo as chaves fortes não identificam, por si sós, qual linha deveria ser descartada e podem representar guias complementares ou eventos administrativos distintos.

Por isso, os números de “excedentes se deduplicasse” são apenas uma medida de impacto potencial. R6 não deve remover registros até existir uma chave transacional estável ou uma verificação adicional de campos administrativos que diferenciem as ocorrências.

```python
# ============================================================
# SOBREPOSIÇÃO ENTRE R1–R5
# ============================================================

matriz_sobreposicao = mascaras_regras.astype("int64").T.dot(
    mascaras_regras.astype("int64")
)
print("Matriz de interseção — diagonal = total da regra; fora da diagonal = interseção do par:")
display(matriz_sobreposicao)

print(f"R1 também em R3: {(mask_r1 & mask_r3).sum():,}")
print(f"R1 também em R2: {(mask_r1 & mask_r2).sum():,}")

uniao_regras = mascaras_regras.any(axis=1)
quantidade_regras_por_registro = mascaras_regras.sum(axis=1)

interpretacoes_regras = {
    "R1": "Sinal econômico forte, mas deriva do alvo; perigosa isoladamente.",
    "R2": "Base fiscal ausente em substância; não invalida a venda por si só.",
    "R3": "Divergência fiscal elevada; usar como alerta, não como prova.",
    "R4": "Regra de escopo temporal, não de qualidade intrínseca.",
    "R5": "Cópia integral redundante; manter uma ocorrência.",
}

linhas_resumo = []
for regra in mascaras_regras.columns:
    mascara = mascaras_regras[regra]
    exclusivos = mascara & quantidade_regras_por_registro.eq(1)
    sobreposicao = mascara & quantidade_regras_por_registro.gt(1)
    linhas_resumo.append(
        {
            "Regra": regra,
            "Registros": int(mascara.sum()),
            "% V2": mascara.mean() * 100,
            "Exclusivos": int(exclusivos.sum()),
            "Sobreposição": int(sobreposicao.sum()),
            "Interpretação": interpretacoes_regras[regra],
        }
    )

resumo_regras_v3 = pd.DataFrame(linhas_resumo)
display(resumo_regras_v3)

print("Combinações observadas entre as regras, excluindo quem não satisfaz nenhuma:")
display(
    mascaras_regras.loc[uniao_regras]
    .value_counts()
    .rename("Registros")
    .reset_index()
)

print("Os exclusivos são a contribuição marginal de cada regra à união completa, sem depender de ordem.")
```

```python
# ============================================================
# IMPACTO HIPOTÉTICO DA UNIÃO — SEM CRIAR OU FILTRAR A V3
# ============================================================

n_v2_atual = len(df_v2)
n_candidatos_uniao = int(uniao_regras.sum())
n_pos_uniao_hipotetico = n_v2_atual - n_candidatos_uniao

impacto_hipotetico = pd.DataFrame(
    {
        "Métrica": [
            "V2 atual",
            "Registros candidatos à remoção pela união R1 | R2 | R3 | R4 | R5",
            "V3 hipotética (somente contagem)",
            "Percentual total preservado desde V1",
            "Percentual total preservado desde a base original de apartamentos",
        ],
        "Valor": [
            n_v2_atual,
            n_candidatos_uniao,
            n_pos_uniao_hipotetico,
            n_pos_uniao_hipotetico / len(df_v1) * 100,
            n_pos_uniao_hipotetico / len(df_apartamentos) * 100,
        ],
    }
)
display(impacto_hipotetico)

print("Nenhum registro foi removido e nenhum objeto df_v3 foi criado.")
```

### Recomendação metodológica antes da V3

A soma simples de R1–R5 seria 1.281, mas a união correta contém **1.098 registros**. Se todas as regras fossem aplicadas, restariam hipoteticamente 62.783 registros: 96,66% da V1 e 79,36% da base original de apartamentos. Essa união integral **não é recomendada**.

A evidência atual sustenta a seguinte decisão:

1. **R5 deve compor a V3**, removendo apenas as 48 cópias excedentes e preservando uma ocorrência de cada linha exata. É a regra com menor risco de eliminar informação real.
2. **R4 deve compor a V3 somente se a população-alvo for definida por `Data de Transação` em 2025.** Nesse caso, seus 669 registros são excluídos por escopo, não por serem inválidos. Como R4 e R5 não se sobrepõem, a aplicação somente dessas duas regras deixaria 63.164 registros, preservando 97,25% da V1 e 79,85% da base original de apartamentos. Se o objetivo for estudar guias pagas em 2025 independentemente da data da transação, R4 não deve ser aplicada.
3. **R1, R2 e R3 não devem compor a V3 como regras isoladas.** R1 deriva do próprio alvo e pode apagar vendas legítimas da cauda inferior. R2 não acrescenta nenhum registro exclusivo: seus 14 casos já pertencem a R1. R3 acrescenta 188 casos exclusivos e usa uma base fiscal que coincide com o próprio valor declarado em 79,86% da V2, portanto não funciona como validação independente do preço.
4. A interseção **`R1 & (R2 | R3)` reúne 168 registros** e é mais defensável como fila de revisão ou quarentena do que qualquer uma das três regras isoladas, pois combina preço/m² extremo com um segundo sinal fiscal. Ainda assim, não deve virar remoção automática antes de validar amostras na fonte e entender os casos de base zero.
5. **R6 não deve compor a V3 neste momento.** A chave ampla `CEP + área + valor + data` reuniu 1.262 registros e captura claramente vendas distintas de unidades semelhantes. As chaves mais fortes `SQL + área + valor + data` e `SQL + cartório + matrícula + valor + data` apontaram somente 22 registros em 11 grupos, mas ainda podem representar republicações ou eventos administrativos legítimos; falta uma chave única da guia/transação para decidir qual ocorrência seria redundante.

As regras mais perigosas para transações legítimas são **R1 e R3**, seguidas por R2. R4 também remove transações legítimas, mas de forma transparente por definição de coorte. A recomendação operacional, antes de criar a V3, é confirmar formalmente se o recorte temporal desejado é a data da transação; com essa confirmação, aplicar apenas `R4 | R5` e manter R1–R3 e R6 como flags auditáveis.

## BASE V3 - CRIAÇÃO E VALIDAÇÃO FORMAL

A V3 parte exclusivamente de `df_v2` e aplica somente estas regras, nesta ordem:

1. **Coorte temporal:** manter apenas registros cuja `Data de Transação` pertence ao ano de 2025. O motivo é definir a população temporal do estudo; registros de outros anos ficam fora do escopo, sem serem classificados como inválidos.
2. **Duplicatas exatas:** depois do recorte temporal, remover somente cópias excedentes de linhas iguais em todas as colunas da V2. A primeira ocorrência na ordem original da V2 é preservada deterministicamente, pois as demais não acrescentam informação.

`Preco_m2` baixo, `Base de Cálculo adotada == 0` e a razão entre valor de transação e base de cálculo não são regras de exclusão. Os casos anteriormente considerados suspeitos por `Preco_m2 < 2.000` permanecem identificados para auditoria. `df_v1` e `df_v2` são preservados, não há treinamento de modelo, análise da cauda superior ou criação de V4 nesta etapa.


```python
# ============================================================
# PRÉ-VALIDAÇÃO DAS DUAS REGRAS DA V3
# ============================================================

registros_esperados_v3 = 63_164

n_v1_antes_v3 = len(df_v1)
n_v2_antes_v3 = len(df_v2)
colunas_v1_antes_v3 = df_v1.columns.tolist()
colunas_v2_antes_v3 = df_v2.columns.tolist()
hash_v1_antes_v3 = pd.util.hash_pandas_object(df_v1, index=True)
hash_v2_antes_v3 = pd.util.hash_pandas_object(df_v2, index=True)

base_v3_trabalho = df_v2.copy()
base_v3_trabalho["_Ordem_original_V2"] = np.arange(
    len(base_v3_trabalho),
    dtype="int64",
)
datas_v3_trabalho = pd.to_datetime(
    base_v3_trabalho["Data de Transação"],
    errors="coerce",
)

# Regra 1: o ano da data da transação define a coorte de 2025.
mask_data_2025 = datas_v3_trabalho.dt.year.eq(2025)
n_removidos_data_fora_2025 = int((~mask_data_2025).sum())
base_v3_2025 = base_v3_trabalho.loc[mask_data_2025].copy()

# Regra 2: todas as colunas da V2 devem ser iguais para caracterizar cópia exata.
base_v3_2025 = base_v3_2025.sort_values("_Ordem_original_V2", kind="mergesort")
mask_duplicata_exata_excedente = base_v3_2025.duplicated(
    subset=colunas_v2_antes_v3,
    keep="first",
)
n_removidos_duplicata_exata = int(mask_duplicata_exata_excedente.sum())
n_v3_prevalidacao = len(base_v3_2025) - n_removidos_duplicata_exata

validacao_previa_v3 = pd.DataFrame(
    {
        "Etapa": [
            "V2 inicial",
            "Após manter somente datas de 2025",
            "Cópias exatas excedentes a remover",
            "V3 projetada",
        ],
        "Registros": [
            len(df_v2),
            len(base_v3_2025),
            n_removidos_duplicata_exata,
            n_v3_prevalidacao,
        ],
    }
)
display(validacao_previa_v3)

if n_v3_prevalidacao != registros_esperados_v3:
    raise RuntimeError(
        "A pré-validação não resultou em 63.164 registros. "
        "Interrompa a criação da V3 e investigue a diferença sem alterar os filtros."
    )
```

```python
# ============================================================
# CRIAÇÃO DA V3 APÓS A PRÉ-VALIDAÇÃO
# ============================================================

indices_removidos_duplicata_exata = base_v3_2025.index[
    mask_duplicata_exata_excedente
]

df_v3 = (
    base_v3_2025.loc[~mask_duplicata_exata_excedente]
    .sort_values("_Ordem_original_V2", kind="mergesort")
    .drop(columns="_Ordem_original_V2")
    .copy()
)

assert len(df_v3) == registros_esperados_v3
assert len(df_v1) == n_v1_antes_v3
assert len(df_v2) == n_v2_antes_v3
assert df_v1.columns.tolist() == colunas_v1_antes_v3
assert df_v2.columns.tolist() == colunas_v2_antes_v3
assert pd.util.hash_pandas_object(df_v1, index=True).equals(hash_v1_antes_v3)
assert pd.util.hash_pandas_object(df_v2, index=True).equals(hash_v2_antes_v3)
```

`Preco_m2` é calculado a partir de valor dividido por área. Portanto, sua associação com área é parcialmente mecânica; os casos extremos exigem conferência documental antes de qualquer regra de tratamento.


```python
# CEP_8: texto de oito algarismos; entradas inválidas ficam ausentes no diagnóstico.
def normalizar_cep(valor):
    if pd.isna(valor):
        return pd.NA
    if isinstance(valor, (int, np.integer)):
        digitos = str(valor)
    elif isinstance(valor, (float, np.floating)):
        if not np.isfinite(valor) or not valor.is_integer():
            return pd.NA
        digitos = str(int(valor))
    else:
        texto = str(valor).strip()
        if re.fullmatch(r"[0-9]{5}-[0-9]{3}", texto):
            texto = texto.replace("-", "")
        if not re.fullmatch(r"[0-9]{7,8}", texto):
            return pd.NA
        digitos = texto
    if not re.fullmatch(r"[0-9]{7,8}", digitos):
        return pd.NA
    return digitos.zfill(8)

assert normalizar_cep(1234567) == "01234567"
assert normalizar_cep("01234-567") == "01234567"
assert pd.isna(normalizar_cep("123"))
assert pd.isna(normalizar_cep("12A34567"))

cep_bruto_ausente = df_area_localizacao["CEP"].isna() | df_area_localizacao["CEP"].astype("string").str.strip().eq("").fillna(False)
df_area_localizacao["CEP_8"] = pd.Series(
    [normalizar_cep(x) for x in df_area_localizacao["CEP"]],
    index=df_area_localizacao.index, dtype="string"
)
cep_invalido = ~cep_bruto_ausente & df_area_localizacao["CEP_8"].isna()
assert df_area_localizacao["CEP_8"].dropna().str.fullmatch(r"[0-9]{8}").all()
contagem_cep = df_area_localizacao["CEP_8"].value_counts()
print("CEP bruto ausente:",int(cep_bruto_ausente.sum()),"| bruto inválido:",int(cep_invalido.sum()))
print("CEP_8 válido:",int(df_area_localizacao["CEP_8"].notna().sum()),"| CEPs únicos:",len(contagem_cep))
print("Imóveis/CEP: distribuição:")
display(contagem_cep.describe(percentiles=[.25,.50,.75,.90,.95,.99]).to_frame("Imóveis_por_CEP"))
print("CEPs com 1, <5 e >=100 imóveis:",int(contagem_cep.eq(1).sum()),int(contagem_cep.lt(5).sum()),int(contagem_cep.ge(100).sum()))
print("Imóveis nos CEPs com <5 e >=100 registros:",int(contagem_cep[contagem_cep.lt(5)].sum()),int(contagem_cep[contagem_cep.ge(100)].sum()))
print("CEPs mais frequentes:")
display(contagem_cep.head(12).to_frame("Imóveis"))
print("Exemplos de CEP bruto inválido:")
display(df_area_localizacao.loc[cep_invalido,["CEP","Bairro"]].head(12))
```

```python
# Bairro: normalização só para comparar grafias; original preservado.
def normalizar_bairro(valor):
    if pd.isna(valor):
        return pd.NA
    texto = re.sub(r"\s+", " ", str(valor).strip())
    if not texto:
        return pd.NA
    texto = "".join(c for c in unicodedata.normalize("NFKD",texto) if not unicodedata.combining(c))
    return texto.upper()

bairro_bruto = df_area_localizacao["Bairro"].astype("string")
bairro_ausente = bairro_bruto.isna() | bairro_bruto.str.strip().eq("").fillna(False)
df_area_localizacao["Bairro_normalizado"] = pd.Series(
    [normalizar_bairro(x) for x in df_area_localizacao["Bairro"]],
    index=df_area_localizacao.index,dtype="string"
)
bairro_valido = bairro_bruto[~bairro_ausente]
bairro_espacos = bairro_valido.ne(bairro_valido.str.strip()) | bairro_valido.str.contains(r"\s{2,}")
bairro_caixa = bairro_valido.ne(bairro_valido.str.upper())
bairro_acento = bairro_valido.map(lambda x: any(unicodedata.combining(c) for c in unicodedata.normalize("NFKD",x)))
variantes_bairro = (df_area_localizacao.dropna(subset=["Bairro_normalizado"])
    .groupby("Bairro_normalizado")["Bairro"].agg(lambda s: sorted(set(s.astype(str)))))
variantes_bairro = variantes_bairro[variantes_bairro.map(len).gt(1)]
genericos = {"NAO INFORMADO","NAO IDENTIFICADO","DESCONHECIDO","SEM INFORMACAO","SEM BAIRRO","BAIRRO","OUTROS","OUTRO","INDEFINIDO","N/A","NA","-","0"}
bairro_generico = df_area_localizacao["Bairro_normalizado"].isin(genericos)
print("Bairro ausente/vazio:",int(bairro_ausente.sum()),"| únicos brutos:",bairro_valido.nunique(),"| únicos normalizados:",df_area_localizacao["Bairro_normalizado"].nunique())
print("Registros com variação de caixa, espaços extras e acentos:",int(bairro_caixa.sum()),int(bairro_espacos.sum()),int(bairro_acento.sum()))
print("Chaves normalizadas com mais de uma grafia:",len(variantes_bairro))
print("Exemplos de variantes:")
display(variantes_bairro.head(15).to_frame("Grafias originais"))
print("Valores genéricos explícitos:",int(bairro_generico.sum()))
print("Bairros mais frequentes:")
display(df_area_localizacao["Bairro_normalizado"].value_counts().head(15).to_frame("Imóveis"))
print("Valores curtos ou aparentemente genéricos:")
display(df_area_localizacao.loc[df_area_localizacao["Bairro_normalizado"].str.len().lt(4).fillna(False) | bairro_generico,"Bairro"].value_counts().head(20).to_frame("Imóveis"))
```

### Estratégia de treino/teste

A V3 contém transações de janeiro a dezembro de 2025. Para simular uma estimativa futura, a avaliação deve usar **datas posteriores** às de treinamento. Uma divisão aleatória misturaria períodos e permitiria ao treino aprender padrões de preços posteriores aos exemplos de teste. A divisão abaixo escolhe a **data inteira** cuja contagem acumulada fica mais próxima de 80%; todos os registros de uma mesma data ficam no mesmo lado. A proporção resultante pode diferir ligeiramente de 80/20. A data é metadado de separação, não feature de `X` nem coluna de `df_modelagem`.

`random_state = 42` fica definido para futuras operações estocásticas. A divisão cronológica em si é determinística e não utiliza sorteio. As transformações que aprenderem parâmetros ou categorias da base deverão ser ajustadas após esta divisão, exclusivamente no treino.


```python
# Divisão temporal, sem usar y para escolher o corte e sem pré-processamento ajustado no teste.
random_state = 42
datas_modelagem = pd.to_datetime(df_v3.loc[df_modelagem.index, "Data de Transação"], errors="coerce")
assert datas_modelagem.notna().all() and datas_modelagem.dt.year.eq(2025).all()

contagem_por_data = datas_modelagem.value_counts().sort_index()
acumulado_treino = contagem_por_data.cumsum().iloc[:-1]
alvo_treino = round(0.80 * len(df_modelagem))
data_corte_treino = (acumulado_treino - alvo_treino).abs().idxmin()
mask_treino = datas_modelagem.le(data_corte_treino)
mask_teste = ~mask_treino

X_train, X_test = X.loc[mask_treino].copy(), X.loc[mask_teste].copy()
y_train, y_test = y.loc[mask_treino].copy(), y.loc[mask_teste].copy()
assert len(X_train) + len(X_test) == len(df_modelagem)
assert len(X_train) == len(y_train) and len(X_test) == len(y_test)
assert X_train.index.intersection(X_test.index).empty
assert datas_modelagem.loc[X_train.index].max() < datas_modelagem.loc[X_test.index].min()
assert abs(len(X_train) / len(X) - 0.80) < 0.005

for nome, base in (("df_v1", df_v1), ("df_v2", df_v2), ("df_v3", df_v3)):
    n_antes, colunas_antes, hash_antes = assinaturas_bases_antes_modelagem[nome]
    assert len(base) == n_antes and base.columns.tolist() == colunas_antes
    assert pd.util.hash_pandas_object(base, index=True).equals(hash_antes)

print("Última data de treino:", datas_modelagem.loc[X_train.index].max().date())
print("Primeira data de teste:", datas_modelagem.loc[X_test.index].min().date())
print("Treino:", len(X_train), f"({len(X_train) / len(X):.2%})")
print("Teste:", len(X_test), f"({len(X_test) / len(X):.2%})")
print("Ausentes em X_train / X_test:")
display(pd.DataFrame({"Treino": X_train.isna().sum(), "Teste": X_test.isna().sum()}))
print("df_v1, df_v2 e df_v3: assinaturas de conteúdo e estrutura preservadas.")
```

## SMARTPRICE AI — BASELINE V1

**Objetivo:** estimar `Valor de Transação (declarado pelo contribuinte)` com as quatro features aprovadas e a divisão cronológica já definida. Usamos uma única regressão Ridge (penalização L2 fixa em 1, sem busca de hiperparâmetros): é um modelo linear simples, com coeficientes interpretáveis, e a penalização estabiliza o ajuste com muitos CEPs categóricos.

O `ColumnTransformer` imputa as numéricas pela mediana e as padroniza. Para as categorias, imputa ausência com uma categoria explícita, agrupa CEPs com menos de cinco ocorrências **no treino** e aplica one-hot encoding. CEPs inéditos no teste usam o grupo infrequente, quando existente; categorias de padrão inéditas são ignoradas pelo codificador. Todo parâmetro do pré-processamento é aprendido dentro do `Pipeline.fit(X_train, y_train)`. `Preco_m2`, o próprio target e as demais colunas ligadas à transação ficam fora de `X`.

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Contrato do dataset e da separação, antes de qualquer fit.
assert len(df_modelagem) == len(X) == len(y) == 63_164
assert len(X_train) == len(y_train) == 50_598
assert len(X_test) == len(y_test) == 12_566
assert X.columns.tolist() == FEATURES_APROVADAS
assert set(FEATURES_NUMERICAS) == {"Área Construída (m2)", "ACC (IPTU)"}
assert set(FEATURES_CATEGORICAS) == {"CEP_8", "Padrão (IPTU)"}
assert set(X.columns).isdisjoint(colunas_leakage)
assert "Preco_m2" not in X.columns and col_target not in X.columns
assert y.name == col_target
assert X_train.index.equals(y_train.index) and X_test.index.equals(y_test.index)
assert X_train.index.intersection(X_test.index).empty
assert datas_modelagem.loc[X_train.index].max() < datas_modelagem.loc[X_test.index].min()
for nome, base in (("df_v1", df_v1), ("df_v2", df_v2), ("df_v3", df_v3)):
    n_antes, colunas_antes, hash_antes = assinaturas_bases_antes_modelagem[nome]
    assert len(base) == n_antes and base.columns.tolist() == colunas_antes
    assert pd.util.hash_pandas_object(base, index=True).equals(hash_antes)

preprocessamento_baseline_v1 = ColumnTransformer(
    transformers=[
        ("numericas", Pipeline([
            ("imputacao", SimpleImputer(strategy="median")),
            ("escala", StandardScaler()),
        ]), FEATURES_NUMERICAS),
        ("cep", Pipeline([
            ("imputacao", SimpleImputer(strategy="constant", fill_value="__AUSENTE__")),
            ("one_hot", OneHotEncoder(min_frequency=5, handle_unknown="infrequent_if_exist")),
        ]), ["CEP_8"]),
        ("padrao", Pipeline([
            ("imputacao", SimpleImputer(strategy="constant", fill_value="__AUSENTE__")),
            ("one_hot", OneHotEncoder(handle_unknown="ignore")),
        ]), ["Padrão (IPTU)"]),
    ]
)
baseline_v1 = Pipeline([
    ("preprocessamento", preprocessamento_baseline_v1),
    ("regressao", Ridge(alpha=1.0, solver="lsqr")),
])
baseline_v1.fit(X_train, y_train)

cep_encoder_v1 = baseline_v1.named_steps["preprocessamento"].named_transformers_["cep"].named_steps["one_hot"]
ceps_treino_v1 = set(X_train["CEP_8"].dropna())
ceps_ineditos_teste_v1 = ~X_test["CEP_8"].isin(ceps_treino_v1)
print("Algoritmo: Ridge (alpha=1, solver=lsqr), sem tuning")
print("CEPs distintos no treino:", len(ceps_treino_v1))
print("CEPs inéditos no teste:", int(ceps_ineditos_teste_v1.sum()))
print("CEPs raros agrupados no fit:", len(cep_encoder_v1.infrequent_categories_[0]))
print("Período de treino:", datas_modelagem.loc[X_train.index].min().date(), "a", datas_modelagem.loc[X_train.index].max().date())
print("Período de teste:", datas_modelagem.loc[X_test.index].min().date(), "a", datas_modelagem.loc[X_test.index].max().date())
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

previsoes_baseline_v1 = baseline_v1.predict(X_test)
real_baseline_v1 = y_test.to_numpy(dtype=float)
erro_assinado_v1 = previsoes_baseline_v1 - real_baseline_v1  # positivo = superestimativa
erro_absoluto_v1 = np.abs(erro_assinado_v1)
assert np.isfinite(previsoes_baseline_v1).all()
assert (real_baseline_v1 > 0).all()

metricas_baseline_v1 = {
    "MAE (R$)": mean_absolute_error(real_baseline_v1, previsoes_baseline_v1),
    "RMSE (R$)": np.sqrt(mean_squared_error(real_baseline_v1, previsoes_baseline_v1)),
    "R²": r2_score(real_baseline_v1, previsoes_baseline_v1),
    "Mediana do erro absoluto (R$)": np.median(erro_absoluto_v1),
    "Erro percentual absoluto mediano (%)": 100 * np.median(erro_absoluto_v1 / real_baseline_v1),
}
avaliacao_baseline_v1 = pd.DataFrame({
    "Data": datas_modelagem.loc[X_test.index].to_numpy(),
    "Valor real (R$)": real_baseline_v1,
    "Valor previsto (R$)": previsoes_baseline_v1,
    "Erro assinado (R$)": erro_assinado_v1,
    "Erro absoluto (R$)": erro_absoluto_v1,
}, index=X_test.index)
print("MÉTRICAS — BASELINE V1 (teste cronológico)")
for nome, base in (("df_v1", df_v1), ("df_v2", df_v2), ("df_v3", df_v3)):
    n_antes, colunas_antes, hash_antes = assinaturas_bases_antes_modelagem[nome]
    assert len(base) == n_antes and base.columns.tolist() == colunas_antes
    assert pd.util.hash_pandas_object(base, index=True).equals(hash_antes)
for nome, valor in metricas_baseline_v1.items():
    print(f"{nome}: {valor:,.4f}" if nome == "R²" or "(%)" in nome else f"{nome}: R$ {valor:,.2f}")
```

```python
# Diagnóstico descritivo; nenhuma observação é excluída do cálculo das métricas.
print("Distribuição do erro assinado (previsto - real), em R$:")
print(avaliacao_baseline_v1["Erro assinado (R$)"].describe(percentiles=[.01, .10, .25, .50, .75, .90, .99]).to_string())
print("\nDez maiores erros absolutos (valores reais x previstos):")
print(avaliacao_baseline_v1.nlargest(10, "Erro absoluto (R$)").to_string())

faixas_valor_v1 = pd.cut(
    avaliacao_baseline_v1["Valor real (R$)"],
    bins=[0, 250_000, 500_000, 1_000_000, 2_000_000, 5_000_000, float("inf")],
    labels=["Até 250 mil", "250–500 mil", "500 mil–1 mi", "1–2 mi", "2–5 mi", "Acima de 5 mi"],
    include_lowest=True,
)
resumo_faixas_baseline_v1 = avaliacao_baseline_v1.groupby(faixas_valor_v1, observed=True).agg(
    Registros=("Valor real (R$)", "size"),
    Real_mediano=("Valor real (R$)", "median"),
    Previsto_mediano=("Valor previsto (R$)", "median"),
    Erro_assinado_mediano=("Erro assinado (R$)", "median"),
    Erro_absoluto_mediano=("Erro absoluto (R$)", "median"),
    Erro_assinado_medio=("Erro assinado (R$)", "mean"),
)
print("\nErros por faixa de valor real (R$):")
print(resumo_faixas_baseline_v1.to_string(float_format=lambda valor: f"{valor:,.2f}"))
print("\nPrevisões negativas:", int((previsoes_baseline_v1 < 0).sum()))
```
